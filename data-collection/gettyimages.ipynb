```python
import requests
from bs4 import BeautifulSoup
import re

# URL to fetch
url = "https://photos.com/featured/1-portrait-of-sadhu-holy-man-kathmandu-nepal-bm-noskowski.html?utm_source=GettyImages&utm_medium=website&utm_campaign=GettyImagesBuyPrint"

# Send a request to the webpage
response = requests.get(url)

# Check if the request was successful (status code 200)
if response.status_code == 200:
    # Parse the HTML content of the page
    soup = BeautifulSoup(response.text, 'html.parser')

    # Extract the og:url meta tag
    og_url_tag = soup.find('meta', attrs={'property': 'og:url'})

    if og_url_tag:
        og_url = og_url_tag.get('content', '')
        
        # Use regular expression to extract the artworkid from the URL
        artworkid_match = re.search(r'artworkid=(\d+)', og_url)

        if artworkid_match:
            artworkid = artworkid_match.group(1)
            print(f"Extracted artworkid: {artworkid}")
        else:
            print("Artwork ID not found in the URL.")
    else:
        print("og:url meta tag not found.")
else:
    print(f"Failed to retrieve the page. Status code: {response.status_code}")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup
import time
import csv

# Set up Selenium WebDriver (update the path to your chromedriver)
service = Service(r'C:\Users\sshak\Downloads\chromedriver-win64\chromedriver-win64\chromedriver.exe')# Up  date path
driver = webdriver.Chrome(service=service)

# URL to the Getty Images search page
url = "https://www.gettyimages.com/search/2/image?license=rf&compositions=closeup%2Clookingatcamera%2Cportrait%2Cabstract&ethnicity=southasian&locations=61036&numberofpeople=one&orientations=vertical&phrase=indian%20portrait&sort=mostpopular"

# Function to extract license URLs'https://www.gettyimages.com/search/2/image-film?family=creative&mediatype=photograp
def extract_license_urls(page_source):
    soup = BeautifulSoup(page_source, 'html.parser')
    meta_tags = soup.findAll("meta", itemprop="acquireLicensePage")
    
    # Open CSV file in append mode to save the URLs
    with open('indian.csv', mode='a', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)
        
        # Write each extracted license URL to the CSV
        for i, meta_tag in enumerate(meta_tags, start=1):
            license_url = meta_tag.get("content")
            writer.writerow([license_url])
            print(f"indian vertical one person{i}: {license_url}")

# Start navigation
driver.get(url)

# Write header to CSV file (if file is empty)
with open('indian.csv', mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['License URL'])  # CSV header

# Explicit wait for the page to load
wait = WebDriverWait(driver, 10)  # Wait up to 10 seconds

while True:
    # Extract data from the current page
    html_content = driver.page_source
    extract_license_urls(html_content)
    
    try:
        # Locate the "Next" button link using its href attribute
        next_button = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, 'a[title="Next page"]')))
        next_page_url = next_button.get_attribute('href')  # Extract the href attribute (next page URL)

        if not next_page_url:
            print("No more pages to navigate.")
            break

        # Navigate to the next page
        driver.get(next_page_url)
        time.sleep(2)  # Wait for the next page to load

    except Exception as e:
        print("No more pages to navigate.")
        break

# Quit the browser
driver.quit()
```

```python
import csv
import os
import requests
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
from concurrent.futures import ThreadPoolExecutor
import asyncio
import aiohttp
import nest_asyncio

# Apply nest_asyncio for interactive environments
nest_asyncio.apply()

# Configure Selenium WebDriver
service = Service(r'C:\Users\sshak\Downloads\chromedriver-win64\chromedriver-win64\chromedriver.exe')  # Update with the correct path to your chromedriver
driver = webdriver.Chrome(service=service)

# Input CSV file containing URLs
input_csv = 'indian.csv'  # Update with your input CSV file name
output_folder = 'indians'  # Folder to save downloaded images

# Ensure output folder exists
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# Function to extract image URL using Selenium and BeautifulSoup
def extract_image_url_from_page(url):
    try:
        driver.get(url)
        driver.implicitly_wait(10)  # Wait for the page to load
        html_content = driver.page_source
        soup = BeautifulSoup(html_content, 'html.parser')

        # Find the correct <source> tag: type="image/jpeg" and no media attribute
        source_tag = soup.find('source', type="image/jpeg", media=False)
        if source_tag and source_tag.has_attr('srcset'):
            return source_tag['srcset']
    except Exception as e:
        print(f"Error extracting image URL from {url}: {e}")
    return None

# Async function to download images concurrently
async def download_image(img_url, image_name):
    try:
        async with aiohttp.ClientSession() as session:
            async with session.get(img_url) as response:
                if response.status == 200:
                    # Save the image to a file
                    file_path = os.path.join(output_folder, image_name)
                    if not os.path.exists(file_path):  # Avoid downloading duplicates
                        with open(file_path, 'wb') as file:
                            file.write(await response.read())
                        print(f"Image downloaded successfully: {file_path}")
                else:
                    print(f"Failed to download image from {img_url}. Status code:", response.status)
    except Exception as e:
        print(f"Error downloading {img_url}: {e}")

# Function to process each row and download images concurrently
async def process_row(i, row):
    source_url = row[0]  # Assuming each row contains one URL
    image_name = f"{i}.jpg"  # Naming images sequentially
    
    # Extract image URL using Selenium
    img_url = extract_image_url_from_page(source_url)
    if img_url:
        # Download the image asynchronously
        await download_image(img_url, image_name)
    else:
        print(f"No valid image found for URL: {source_url}")

# Main function to handle CSV reading and image downloading
async def main():
    with open(input_csv, mode='r', newline='', encoding='utf-8') as infile:
        reader = csv.reader(infile)
        tasks = []
        for i, row in enumerate(reader, start=1):  # Start numbering from 1
            tasks.append(process_row(i, row))
        await asyncio.gather(*tasks)  # Run tasks concurrently

# Run the main function
if __name__ == "__main__":
    loop = asyncio.get_event_loop()
    loop.run_until_complete(main())
    driver.quit()
    print("Image downloading completed.")
```

```python
import csv
import requests
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup

# Configure Selenium WebDriver
service = Service(r'C:\Users\sshak\Downloads\chromedriver-win64\chromedriver-win64\chromedriver.exe')  # Update with the correct path to your chromedriver
driver = webdriver.Chrome(service=service)

# Input CSV file containing URLs
input_csv = 'indian.csv'  # Update with your input CSV file name
output_folder = 'indians'  # Folder to save downloaded images

# Ensure output folder exists
import os

# Ensure output folder exists
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# Function to extract and download an image from a URL
def download_image_from_url(url, image_name):
    try:
        driver.get(url)
        driver.implicitly_wait(10)  # Wait for the page to load
        
        # Get the page source after it's fully loaded
        html_content = driver.page_source
        soup = BeautifulSoup(html_content, 'html.parser')
        
        # Find the correct <source> tag: type="image/jpeg" and no media attribute
        source_tag = soup.find('source', type="image/jpeg", media=False)
        if source_tag and source_tag.has_attr('srcset'):
            img_url = source_tag['srcset']
            
            # Download the image
            response = requests.get(img_url)
            if response.status_code == 200:
                # Save the image to a file
                file_path = os.path.join(output_folder, image_name)
                with open(file_path, 'wb') as file:
                    file.write(response.content)
                # print(f"Image downloaded successfully: {file_path}")
            else:
                print(f"Failed to download image from {url}. Status code:", response.status_code)
        else:
            print(f"No valid <source> tag with desired attributes found for URL: {url}")
    except Exception as e:
        print(f"Error processing {url}: {e}")

# Read the input CSV file and process each URL
with open(input_csv, mode='r', newline='', encoding='utf-8') as infile:
    reader = csv.reader(infile)
    for i, row in enumerate(reader, start=1):
        source_url = row[0]  # Assuming each row contains one URL
        image_name = f"{i}.jpg"  # Naming images sequentially
        # print(f"Processing URL {i}: {source_url}")
        download_image_from_url(source_url, image_name)

# Close the Selenium WebDriver
driver.quit()

print("Image downloading completed.")
```
